```python
city1 = "  bengaluru"
city2 = "BENGALURU "
city3 = "Bengaluru"
city4 = " chennai "
city5 = "MUMBAI"

clean_city1 = city1.title().strip()
clean_city2 = city2.title().strip()
clean_city3 = city3.title().strip()
clean_city4 = city4.title().strip()
clean_city5 = city5.title().strip()

print(f"{clean_city1}, {clean_city2}, {clean_city3}, {clean_city4}, {clean_city5}")

print(f"Are city1 and city2 the same after cleaning? {clean_city1 == clean_city2}")
```

```python
fixed_cost = 5000
cost_per_learner = 200
learners = 30
total_cost = fixed_cost + cost_per_learner * learners

print(total_cost)

over_10000 = total_cost > 10000
print(over_10000)

print(f"Within the budget of 12000? {total_cost <= 12000}")
```
