```python
## Statistics: mean, median and mode

import statistics

defects = [4, 7, 3, 7, 9, 5, 7, 4, 35]

print(f"Mean: {statistics.mean(defects)}")
```

```python
print(f"Median: {statistics.median(defects)}")
```

```python
print(f"Mode: {statistics.mode(defects)}")
```
