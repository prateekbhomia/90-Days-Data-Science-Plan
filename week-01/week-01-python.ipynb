```python
learner_quiz_score = int(input("learner_quiz_score (0-100): "))
learner_name = input("Learner Name: ")

if learner_quiz_score >= 85:
    grade = "Distinction"
elif learner_quiz_score >= 70:
    grade = "Merit"
elif learner_quiz_score >= 50:
    grade = "Pass"
else:
    grade = "Needs Support"

print(f"{learner_name} scored {learner_quiz_score}: {grade}")
```
